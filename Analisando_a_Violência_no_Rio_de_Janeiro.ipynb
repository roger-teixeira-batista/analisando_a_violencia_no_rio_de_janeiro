# Analisando a Violência no Rio de Janeiro

O Rio de Janeiro é considerado por muitos a Cidade Maravilhosa.
Com seus pontos turísticos inigualáveis e culinária ímpar, atrai turistas de outros cantos do país e de todo canto do mundo.
Como todo lugar, possui seus problemas. A violência urbana sempre fez parte da vida cotidiana de seus habitantes, violência essa que acaba algumas vezes sobrepondo a beleza desse lugar paradisíaco.
Nesse projeto, vamos analisar dados de violência na cidade disponibilizados pela Polícia Militar do Rio de Janeiro, uma das instituições mais respeitadas do Brasil, e iremos descobrir informações importantes à cerca dessa mazela que insiste em perturbar a população.

## Informações Estatísticas da Violência no Rio de Janeiro

Nessa etapa, iremos verificar como os nossos dados estão distribuídos por meio de ferramentas estatísticas.

Nessa fase, iremos executar as seguintes etapas:

* Imprimir o resumo estatístico do DataFrame
* Encontrar as médias das seguintes variáveis (colunas):
  * `roubo_veiculo`
  * `furto_veiculos`
  * `recuperacao_veiculos`
* Calcular qual a porcentagem de carros recuperados em relação aos carros roubados + carros furtados:
  * $\frac{\text{roubo_veiculo} + \text{furto_veiculos}}{ \text{recuperacao_veiculos}}$
* Encontrar os valores máximos e mínimos da coluna `hom_doloso`


```
# Imprimir o resumo estatístico do DataFrame
df.describe()
```

Ocorrem em média **2 vezes** mais homícidios **dolosos** do que homicídios **culposos**.

Ao compararmos as **médias** de **policiais militares mortos em serviço** e **policiais civis mortos em serviço** percebemos que **morrem 6 vezes mais policiais militares do que policiais civis** na cidade do Rio de Janeiro.

```
# média da variável roubo_veiculo
media_roubo = df.roubo_veiculo.mean()
media_roubo
```

Uma **média de 2448,69 veículos roubados** na cidade do Rio de Janeiro desde 1991.

```
# média da variável furto_veiculos
media_furto = df.furto_veiculos.mean()
media_furto
```

Uma **média de 1674,61 veículos furtados** na cidade do Rio de Janeiro desde 1991.

```
# média da variável recuperacao_veiculos
media_recuperado = df.recuperacao_veiculos.mean()
media_recuperado
```

Uma **média de 1772,23 veículos recuperados** na cidade do Rio de Janeiro desde 1991.

```
# percentual de carros roubados e furtados em relação a carros recuperados
((media_roubo + media_furto) / media_recuperado)
```

Existem aproximadamente **2,33 veículos roubados ou furtados não recuperados para cada veículo recuperado**.

```
# Encontrar os valores máximos e mínimos da coluna hom_doloso
max = df.hom_doloso.max()
min = df.hom_doloso.min()

amplitude = max - min

print(f"Máximo: {max}")
print(f'Mínimo: {min}')
print(f'Amplitude: {amplitude}')
```
